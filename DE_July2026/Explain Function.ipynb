```python
from pyspark.sql import SparkSession  #main entry poin to spark

spark = SparkSession.builder \
    .appName("GoToSpark-Test") \
    .getOrCreate()

print(spark.version)
```

```python
import os
print(os.getcwd())
```

```python
df = spark.read.csv(
    "/workspace/data/superstore.csv",
    header = True,
    inferSchema = True
)
    
```

```python
df.show(5)
```

```python
df.printSchema()
```

```python
df.describe()
```

```python
df.describe().show()
```

```python
df.select("Category","Sub-Category", "Sales").show()
```

```python
from pyspark.sql import functions as F
```

```python
df.select(F.col("Category"), F.col("Sales")).show(5)
```

```python
df.filter(F.col("Sales") > 1000).show(5)
```

```python
display(df.filter(F.col("Sales") > 1000))
```

```python
df.select("Category", "Region","Sales").filter(F.col("Sales") >1000).show(5)
```

```python
df.select("Category", "Region","Sales").filter(F.col("Category")=="Furniture").show(5)
```

```python
df.select("Category", "Region","Sales").filter((F.col("Category")=="Furniture") & (F.col("Region")=="South")).show(5)
```

```python
df.groupby("Region").count().show()
```

```python
#select Region, Sum(Sales) as Total_Sales
From Tablename
Group by Region
```

```python
df.groupBy("Region").agg(F.sum("Sales")).show()
```

```python
df.groupBy("Region").agg(F.sum("Sales").alias("Total_Sales")).show()
```

```python
df.groupBy("Region").agg(F.round(F.sum("Sales"),2).alias("Total_Sales")).show()
```

```python
df.groupBy("Region").agg(F.round(F.sum("Sales"),2).alias("Total_Sales")).orderBy(F.col("Total_Sales").desc()).show()
```

```python
df_margin = df.withColumn("Profit_Margin",(F.col("Profit")/F.col("Sales"))*100).select("Sales", "Profit", "Profit_Margin")
```

```python
df_margin.show(5)
```

```python
df_rename =  (df
    .withColumnRenamed("Order Date","Order_Date")
    .withColumnRenamed("Ship Date", "Ship_Date")
             )
```

```python
df_rename.show(5)
```

```python
df_range = df.withColumn(
                            "Sales_Range",
                               F.when(F.col("Sales") <100, "Low")
                                .when(F.col("Sales") <500, "Medium")
                                .when(F.col("Sales") <1000, "High")
                                .otherwise("Very High")
                              )
```

```python
df_salesrange = df_range.select("Sales", "Sales_Range").show(10)
```

```python
df.filter(F.col("Sales") > 1000).explain()
```

```python
df.filter(F.col("Sales") > 1000).explain("formatted")
```

#Spark reads CSv file (Spark CSV reader)
ReadSchema: struct<Region:string,Sales:double>
# Column Pruning: read only the required columns based on select()

Without filter --> Reads all the data then apply filter Region ==South
#Pushed Filters (predicate pushdown): first it will apply the filter (Region = South ) then Spark reads the data

Null == South 

```python
df.filter(F.col("Region")=="South").select("Region", "Sales").explain("formatted")
```

```python
df.filter(F.col("Region")=="South").groupBy("Category").agg(F.sum("Sales").alias("Total_Sales")).explain("formatted")
```

```python
df.filter(F.col("Region")=="South").groupBy("Category").agg(F.sum("Sales").alias("Total_Sales")).explain("cost")
```
