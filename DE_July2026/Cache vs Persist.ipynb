```python
from pyspark.sql import SparkSession  #main entry poin to spark

spark = SparkSession.builder \
    .appName("GoToSpark-Test") \
    .getOrCreate()

print(spark.version)
```

```python
df = spark.read.csv(
    "/workspace/data/superstore.csv",
    header = True,
    inferSchema = True
)
    
```

```python
from pyspark.storagelevel import StorageLevel
from pyspark.sql import functions as F
import time
```

```python
df_test = df.select("Category", "Region","Sales")
df_test.count()
```

```python
start = time.time()
df_test.count()
print("Run without cache:", time.time() -start, "seconds")
```

```python
start = time.time()
df_test.show(5)
print("Run without cache:", time.time() -start, "seconds")
```

```python
df_test.cache()
```

```python
start = time.time()
df_test.count()
print("Run with cache:", time.time() -start, "seconds")
```

```python
start = time.time()
df_test.show(5)
print("Run  cache:", time.time() -start, "seconds")
```
