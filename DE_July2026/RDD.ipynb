```python
from pyspark.sql import SparkSession  #main entry poin to spark

spark = SparkSession.builder \
    .appName("GoToSpark-Test") \
    .getOrCreate()

print(spark.version)
```

```python
print(spark.sparkContext)  # connection between our application and the spark cluster
```

```python
numbers = spark.range(100)
```

```python
numbers.count()
```

```python
lst = [1,2,3,4,5]
print(lst)
```

```python
type(lst)
```

```python
sparkcon = spark.sparkContext
```

```python
rdd1 =sparkcon.parallelize(lst)
```

```python
rdd1.collect()  //Collect() is an action
```

```python
rdd1.getNumPartitions()
```

```python
rdd1.glom().collect()
```

We can Change Number of Partitions
Coalesce()   and  repartition()

```python
rdd2 =rdd1.coalesce(2)   #reduce the num of partitions
```

```python
rdd2.getNumPartitions()
```

```python
rdd2.glom().collect()
```

```python
rdd3= rdd1.repartition(6)  # increase the number of Partitions 8+6
```

```python
rdd3.getNumPartitions()
```

```python
rdd3.glom().collect()
```
