```python
from pyspark.sql import SparkSession  #main entry poin to spark

spark = SparkSession.builder \
    .appName("GoToSpark-Test") \
    .getOrCreate()

print(spark.version)
```

```python
df = spark.read.csv(
    "/workspace/data/superstore.csv",
    header = True,
    inferSchema = True
)
    
```

```python
df.createOrReplaceTempView("superstore")
```

```python
sql_result = spark.sql(""" Select * From superstore """)
```

```python
sql_result.show(5)
```

```python
spark.sql(""" Select 'Customer ID', Category, Sales From superstore """).show(5)
```

```python
df.select(
    "Customer ID",
    "Category",
    "Sales"
).show(5)
```

```python
spark.sql(""" Select Distinct Category From superstore """).show()
```

```python
df.select("Category").distinct().show()
```
