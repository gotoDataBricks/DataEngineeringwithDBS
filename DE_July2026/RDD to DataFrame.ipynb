```python
from pyspark.sql import SparkSession  #main entry poin to spark

spark = SparkSession.builder \
    .appName("GoToSpark-Test") \
    .getOrCreate()

print(spark.version)
```

```python

data = [
    (1, "John", 100000),
    (2, "Adam", 200000),
    (3, "Sam", 300000)
]
```

```python
sparkcon = spark.sparkContext
```

```python
rdd=sparkcon.parallelize(data)
```

```python
rdd.collect()
```

```python
df = rdd.toDF( 
    ["id", "name", "salary" ]
)
```

```python
display(df)
```

```python
df2 = spark.createDataFrame(data, ["id", "name", "salary"] )
```

```python
display(df2)
```

```python
from pyspark.sql.types import (StructType, StructField , IntegerType, StringType)

```

```python
schema = StructType ([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("salary", IntegerType(), True)
])
```

```python
df3 = spark.createDataFrame(rdd, schema)
```

```python
display(df3)
```
