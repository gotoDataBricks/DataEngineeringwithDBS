#transform Orders Data (Json string to JSON Object)

```sql
Select * FROM ecomm.bronze.v_orders
```

```sql
Select value,
regexp_replace(value,'"order_date": (\\d{4}-\\d{2}-\\d{2})','"order_date": "\$1"') as new_value
from ecomm.bronze.v_orders
```

```sql
Select value,
regexp_replace(
  value,
  '"order_date": ?\\d{4}-\\d{2}-\\d{2}',
  concat('"order_date": "', regexp_extract(value, '"order_date": ?(\\d{4}-\\d{2}-\\d{2})', 1), '"')
) as new_value
from ecomm.bronze.v_orders
```

```sql
Select value,
regexp_replace(value,'"order_date": (\\d{4}-\\d{2}-\\d{2})','"order_date": "\$1"') as new_value
from ecomm.bronze.v_orders
```

```sql
CREATE or REPLACE TEMPORARY VIEW tv_orders_fix 
AS
Select value,
regexp_replace(value,'"order_date": (\\d{4}-\\d{2}-\\d{2})','"order_date": "\$1"') as new_value
from ecomm.bronze.v_orders;
```

```sql
Select schema_of_json(new_value) as schema,
new_value
from tv_orders_fix
limit 1;
```

```sql
SELECT from_json(new_value, 'STRUCT<customer_id: BIGINT, items: ARRAY<STRUCT<category: STRING, details: STRUCT<brand: STRING, color: STRING>, item_id: BIGINT, name: STRING, price: BIGINT, quantity: BIGINT>>, order_date: STRING, order_id: BIGINT, order_status: STRING, payment_method: STRING, total_amount: BIGINT, transaction_timestamp: STRING>') as json_value,
new_value
from tv_orders_fix
```

```sql
Create table if not exists ecomm.silver.orders_json
as
SELECT from_json(new_value, 'STRUCT<customer_id: BIGINT, items: ARRAY<STRUCT<category: STRING, details: STRUCT<brand: STRING, color: STRING>, item_id: BIGINT, name: STRING, price: BIGINT, quantity: BIGINT>>, order_date: STRING, order_id: BIGINT, order_status: STRING, payment_method: STRING, total_amount: BIGINT, transaction_timestamp: STRING>') as json_value
from tv_orders_fix;
```

```sql
Select json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
array_distinct(json_value.items) as items
from ecomm.silver.orders_json
```

```sql
Select json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
explode(array_distinct(json_value.items)) as item
from ecomm.silver.orders_json
```

```sql
create or replace temporary view tv_orders_exploded
as
Select json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
explode(array_distinct(json_value.items)) as item
from ecomm.silver.orders_json
```

```sql
Select order_id,
order_date,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
item
from tv_orders_exploded
```

```sql
Select order_id,
order_date,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
item.item_id,
item.name,
item.price,
item.quantity,
item.category,
item.details.brand,
item.details.color
from tv_orders_exploded
```

```sql
create table if not exists ecomm.silver.orders
as
Select order_id,
order_date,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
item.item_id,
item.name,
item.price,
item.quantity,
item.category,
item.details.brand,
item.details.color
from tv_orders_exploded;
```
