#Create a Catalog

```sql
Create catalog  if not exists ktproject
managed location 'abfss://ktproject@azdbsktdejulstorageac.dfs.core.windows.net/'
comment 'Catalog for the ktproject'
```

```sql
create schema if not exists ktproject.delta_lake
managed location 'abfss://ktproject@azdbsktdejulstorageac.dfs.core.windows.net/delta_lake'
comment 'Schema for the ktproject.delta_lake'
```

###1.Create a Delta lake Table

```sql
create table if not exists ktproject.delta_lake.companies
(
    company_name String,
    founded_date Date,
    country String
);
```

```sql
desc extended ktproject.delta_lake.companies
```

#Insert some sample data

```sql
Insert into ktproject.delta_lake.companies
Values ("Apple", "1976-04-01", "USA");
```

```sql
Insert into ktproject.delta_lake.companies
values("Microsoft", "1975-04-01", "USA"),
("Google", "1998-04-01", "USA"),
("Alibaba", "1999-04-01", "China");
```

History and Time Travel

```sql
describe history ktproject.delta_lake.companies;
```

```sql
select * from ktproject.delta_lake.companies
```

2. Query Data from a Specific Version

```sql
Select * from ktproject.delta_lake.companies
version as of 2;
```

```sql
Select * from ktproject.delta_lake.companies
timestamp as of '2026-08-07T02:20:22.000+00:00'
```

Restore Data in the Table

```sql
Restore table ktproject.delta_lake.companies to version as of 1;
```
