```sql
Select refund_id,
payment_id,
cast(date_format(refund_timestamp, 'yyyy-MM-dd') as Date) as refund_date,
date_format(refund_timestamp, 'HH-mm-ss') as refund_time,
refund_amount,
Split(refund_reason,':')[0] as refund_reason,
Split(refund_reason,':')[1] as refund_by
from ecomm.bronze.refunds
```

create table into Silver Schema

```sql
CREATE TABLE ecomm.silver.refunds AS
Select refund_id,
payment_id,
cast(date_format(refund_timestamp, 'yyyy-MM-dd') as Date) as refund_date,
date_format(refund_timestamp, 'HH-mm-ss') as refund_time,
refund_amount,
Split(refund_reason,':')[0] as refund_reason,
Split(refund_reason,':')[1] as refund_by
from ecomm.bronze.refunds
```
